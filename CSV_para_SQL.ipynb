# Convertendo CSV para SQL

Esse script assume que o arquivo CSV tem as seguintes colunas

- coluna1
- coluna2
- coluna3

Cria uma tabela chamada meus_dados, que você pode adpatar para a sua realidade no banco SQLite e insere os dados CSV nessa tabela.

Você pode substituir **'dados.csv'** pelo nome do seu arquivo CSV e **'meu_banco.db'** pelo nome do seu banco de dados SQLite.

**Lembre de criar seu banco de dados SQLite antes de executar o script**

```python
import csv
import sqlite3
```

```python
# definindo os nomes dos arquivos

ARQUIVO_CSV = 'dados.csv'
NOME_BANCO = 'meu_banco.db'

# conexão com o banco
con = sqlite3.connect(NOME_BANCO)
cursor = con.cursor()

# leitura do arquivo CSV
with open(ARQUIVO_CSV, 'r') as arquivo:
    leitor = csv.reader(arquivo)

    next(leitor)  # pula a primeira linha (cabeçalho)

    for linha in leitor:
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS meus_dados
            (id INTEGER PRIMARY KEY,
            coluna1 TEXT,
            coluna2 TEXT,
            coluna3 TEXT)
        ''')

        cursor.execute('''
            INSERT INTO meus_dados (coluna1, coluna2, coluna3)
            VALUES (?, ?, ?)
        ''', linha)

    # confirma as alterações e fecha a conexão
    con.commit()
    con.close()

print('Conversão realizado com sucesso')
```
